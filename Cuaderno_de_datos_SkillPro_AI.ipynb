# 1. Introducción y planteamiento del problema

SkillPro AI es un mentor de desarrollo profesional basado en inteligencia artificial, diseñado para ayudar a estudiantes y profesionales a identificar sus competencias, reconocer sus brechas de conocimiento y convertir sus objetivos laborales en planes de aprendizaje concretos y verificables.

El proyecto busca orientar especialmente a estudiantes de Ingeniería Informática que desean prepararse para su primer empleo en el área tecnológica, pero que necesitan identificar qué conocimientos fortalecer, qué proyectos desarrollar y cómo demostrar sus capacidades ante posibles empleadores.

**Problema abordado:** Las personas que comienzan su carrera profesional pueden tener dificultades para identificar las competencias que necesitan fortalecer, priorizar sus necesidades de aprendizaje y organizar su tiempo para desarrollar proyectos que demuestren sus conocimientos. SkillPro AI busca orientar este proceso mediante diagnósticos, análisis de brechas, planes de acción e indicadores de progreso.

**Usuarios previstos:** Estudiantes universitarios, desarrolladores junior, profesionales que desean actualizar sus conocimientos y personas interesadas en cambiar o iniciar su trayectoria profesional.

**Alcance:** El proyecto proporciona orientación, planificación inicial, recomendaciones de aprendizaje y criterios para evaluar el progreso. No garantiza la contratación laboral ni sustituye la experiencia práctica.

**Caso de estudio:** Una estudiante de Ingeniería Informática con conocimientos básicos de Python, lógica de programación y SQL, sin proyectos completos en GitHub, que desea prepararse para un primer empleo como desarrolladora junior.

**Horizonte de planificación:** Cuatro semanas, con ocho horas de dedicación semanal, equivalentes a 32 horas. La meta profesional se plantea para aproximadamente dos meses, por lo que será necesario realizar seguimiento adicional después del primer periodo.

# 3. Objetivos del proyecto




## 3.1. Objetivo general

Diseñar, documentar y evaluar SkillPro AI como un mentor de desarrollo profesional basado en inteligencia artificial que aplique la metodología D.A.P.E. para proporcionar orientación específica, realista y adaptada al perfil, las necesidades y la disponibilidad de cada usuario.




## 3.2. Objetivos específicos

1. Identificar las necesidades de desarrollo profesional y las competencias relacionadas con el objetivo del usuario.
2. Diseñar una metodología de diagnóstico, análisis de brechas, planificación y evaluación.
3. Definir las instrucciones y el comportamiento esperado de SkillPro AI mediante el archivo SKILL.md.
4. Configurar y probar el asistente en Gemini mediante casos de uso representativos.
5. Evaluar la calidad de las respuestas utilizando criterios observables.
6. Identificar oportunidades de mejora a partir de los resultados reales de las pruebas.
7. Documentar el proceso de desarrollo y las evidencias obtenidas.




## 3.3. Alcance funcional

SkillPro AI contempla las siguientes funciones:

- Diagnóstico de habilidades y objetivos profesionales.
- Identificación y priorización de brechas de conocimiento.
- Elaboración de planes de aprendizaje personalizados.
- Recomendación de proyectos prácticos para el portafolio.
- Definición de objetivos SMART e indicadores KPI.
- Orientación sobre Git, GitHub, documentación y evaluación del progreso.
- Comunicación profesional, clara y adaptada al nivel del usuario.

**Fuera de alcance:** Garantizar la obtención de un empleo, afirmar que se adquirió una competencia sin evidencia o presentar recomendaciones no verificadas como hechos.

# 9. Conclusiones y reflexión




## 9.1. Necesidad atendida

SkillPro AI busca facilitar la planificación del desarrollo profesional de estudiantes y principiantes en tecnología. Su enfoque consiste en identificar las competencias actuales, priorizar las brechas de conocimiento y transformarlas en acciones concretas que puedan evaluarse mediante indicadores y evidencias.




## 9.2. Resultados iniciales

Las dos respuestas analizadas mostraron que SkillProIA puede estructurar un diagnóstico y un análisis de brechas a partir de la información proporcionada. También relacionó las recomendaciones con el objetivo de construir una API REST y con la disponibilidad de ocho horas semanales.




## 9.3. Limitaciones identificadas

La evaluación inicial se limita a dos casos conversacionales. No se ha demostrado todavía la viabilidad del plan mediante su ejecución, la corrección técnica de todos los pasos, la construcción de TaskAPI ni la publicación del repositorio.

También es necesario verificar la correspondencia entre las instrucciones del Gem y la versión definitiva de SKILL.md.




## 9.4. Mejoras propuestas

1. Seleccionar un único framework para el proyecto técnico.
2. Incorporar criterios de aceptación medibles para cada entregable.
3. Ampliar los KPI para incluir pruebas automatizadas y funcionalidades verificadas.
4. Ejecutar pruebas con perfiles y objetivos diferentes.
5. Recopilar comentarios de la usuaria sobre la claridad y utilidad de las respuestas.
6. Revisar las recomendaciones técnicas con documentación fiable.
7. Registrar los cambios en las instrucciones y repetir las pruebas afectadas.




## 9.5. Conclusión provisional

SkillPro AI presenta una propuesta organizada para orientar el desarrollo profesional mediante la metodología D.A.P.E. Las pruebas P01 y P02 proporcionan evidencia inicial de que puede generar un diagnóstico, priorizar brechas y sugerir acciones relacionadas con una meta concreta.

Sin embargo, los resultados son preliminares. La calidad final del proyecto dependerá de pruebas adicionales, la validación de las recomendaciones, la ejecución de las actividades propuestas y la documentación de resultados reales.

**Siguiente paso:** ejecutar la prueba P03, revisar el resultado y registrar si el plan SMART respeta las ocho horas semanales y contiene entregables verificables.
